```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
print(cv2.__version__)
print("Lets start working together with open cv as well")
```

```python
img = cv2.imread('images.jpeg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
#bgr to rbg is done because the matplotlib understands rgb like the numpy library
plt.imshow(img)
print(img.size)
print(img.shape)
plt.axis("off")
plt.show()
```

```python
img_Red = img[:,:,0]
img_Green = img[:,:,1]
img_Blue = img[:,:,2]
new_img = np.hstack((img_Red,img_Green,img_Blue))
plt.imshow(new_img)
```

```python
#Resizing using image cv 
img_resize = cv2.resize(img,(800,800))
resized_rgb = cv2.cvtColor(img_resize, cv2.COLOR_BGR2RGB)
plt.imshow(resized_rgb)
```

```python
# to half the size 
img_half = cv2.resize(img , (img.shape[1]//2,img.shape[0]//2))
img_new = cv2.cvtColor(img_half,cv2.COLOR_BGR2RGB)
plt.imshow(img_new)
print(img_new.shape)
    
```

```python
img_flip = cv2.flip(img,-1) # 0 is vertically flipped # 1 is horizontally flipped #-1 is both verically and horizontally  
img_flipped = cv2.cvtColor(img_flip,cv2.COLOR_BGR2RGB)
plt.imshow(img_flipped)
```

```python
img_crop = img[50:300,20:500] #basic numpy slicing 
plt.imshow(img_crop)
```
